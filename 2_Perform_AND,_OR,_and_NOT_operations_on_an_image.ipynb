```python
# Import Libraries

import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
 # Upload One Image
from google.colab import files
uploaded = files.upload()
```

```python
# Read Image in Grayscale
filename = next(iter(uploaded.keys()))
img = cv2.imread(filename, 0)

if img is None:
    raise ValueError("Image could not be loaded.")
```

```python
# Get Image Size
h, w = img.shape
```

```python
# Create AND Mask
and_mask = np.zeros((h, w), dtype=np.uint8)
```

```python
# Select a rectangular area
x1 = int(w * 0.40)
x2 = int(w * 0.72)
y1 = int(h * 0.02)
y2 = int(h * 0.34)

and_mask[y1:y2, x1:x2] = 255
```

```python
#  Perform AND Operation
and_result = cv2.bitwise_and(img, and_mask)
```

```python
#  Create OR Mask
or_mask = np.full((h, w), 255, dtype=np.uint8)

or_mask[y1:y2, x1:x2] = 0
```

```python
#  Perform OR Operation
or_result = cv2.bitwise_or(img, or_mask)
```

```python
#  Perform NOT Operation
not_result = cv2.bitwise_not(img)
```

```python
# Display AND and OR Results
plt.figure(figsize=(14, 9))

plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(and_mask, cmap='gray')
plt.title("AND Mask")
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(and_result, cmap='gray')
plt.title("AND Result")
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(or_mask, cmap='gray')
plt.title("OR Mask")
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(or_result, cmap='gray')
plt.title("OR Result")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
#  Display NOT Result
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(not_result, cmap='gray')
plt.title("NOT Result")
plt.axis('off')

plt.tight_layout()
plt.show()
```
