```python
1# Translation
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('/content/A-White-Flower-.png')
image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
h,w = image.shape[:2]
tx = int(input ("Enter translation in X direction:"))
ty = int(input ("Enter translation in Y direction:"))
M = np.float32([
 [1,0,tx],
 [0,1,ty]

])
translated = cv2.warpAffine(image,M,(w,h))
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(translated)
plt.title("translated Image")
plt.axis("off")
plt.show()
```

```python
# Scaling
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('/content/A-White-Flower-.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
h, w = image.shape[:2]
sx = 3.5
sy = 5.5
M = np.float32([
 [sx, 0, 0],
 [0, sy, 0]
])
scaled = cv2.warpAffine(
 image,
 M,
 (int(w * sx), int(h * sy))
)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(scaled)
plt.title("Scaled Image")
plt.axis("off")
plt.show()
```

```python
# Rotation
import cv2
import matplotlib.pyplot as plt
image = cv2.imread('/content/A-White-Flower-.png')
image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
h,w = image.shape[:2]
center = (w // 2,h // 2)
angle = 65
M = cv2.getRotationMatrix2D(center,angle,1.0)
rotated = cv2.warpAffine(image,M,(w,h))
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(rotated)
plt.title("Rotated Image-65")
plt.axis("off")
plt.show()
```

```python
# Reflection
import cv2
import matplotlib.pyplot as plt
image = cv2.imread('/content/A-White-Flower-.png')
image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
reflected_x = cv2.flip(image,0)
reflected_y = cv2.flip(image,1)
plt.figure(figsize=(15,5))
plt.subplot(1,3,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(reflected_x)
plt.title("Reflection about X-axis")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(reflected_y)
plt.title("Reflection about Y-axis")
plt.axis("off")

plt.show()
```

```python
# Combine
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('/content/A-White-Flower-.png')
image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
h,w = image.shape[:2]
tx, ty = 100, 120
M_translate = np.float32([
    [1, 0, tx],
    [0, 1, ty]
])
translated = cv2.warpAffine(image, M_translate, (w, h))
center = (w // 2, h // 2)
angle = 65
M_rotate = cv2.getRotationMatrix2D(center, angle, 1.0)
rotated = cv2.warpAffine(image, M_rotate, (w, h))
reflected_x = cv2.flip(image, 0)
reflected_y = cv2.flip(image, 1)
sx2, sy2 = 3.5, 5.5
combined_scaled = cv2.resize(image, None, fx=sx2, fy=sy2)

h2, w2 = combined_scaled.shape[:2]
center2 = (w2 // 2, h2 // 2)
M_combined_rotation = cv2.getRotationMatrix2D(center2, 30, 1.0)
combined_rotated = cv2.warpAffine(combined_scaled, M_combined_rotation, (w2, h2))

tx2, ty2 = 50, 30
M_combined_translation = np.float32([
    [1, 0, tx2],
    [0, 1, ty2]
])
combined_final = cv2.warpAffine(
    combined_rotated,
    M_combined_translation,
    (w2 + tx2, h2 + ty2)
)
titles = [
    "Original", "Translated", "Scaled",
    "Rotated 65°", "Reflect X-axis", "Reflect Y-axis"
    "Combined: Final"
]
images = [
    image, translated,
    rotated, reflected_x, reflected_y,
    combined_final
]

plt.figure(figsize=(20, 10))
for i, (title, img) in enumerate(zip(titles, images), start=1):
    plt.subplot(2, 4, i)
    plt.imshow(img)
    plt.title(title)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
