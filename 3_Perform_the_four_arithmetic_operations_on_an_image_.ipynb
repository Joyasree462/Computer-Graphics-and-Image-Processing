```python
# Import Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#  Upload One Image
from google.colab import files
uploaded = files.upload()
```

```python
#  Read Image
filename = next(iter(uploaded.keys()))
img = cv2.imread(filename, 0)

if img is None:
    raise ValueError("Image could not be loaded.")
```

```python
#  Addition
addition = np.clip(img.astype(float) + 50, 0, 255).astype(np.uint8)
```

```python
#  Subtraction
subtraction = np.clip(img.astype(float) - 50, 0, 255).astype(np.uint8)
```

```python
#  Multiplication
multiplication = np.clip(img.astype(float) * 1.5, 0, 255).astype(np.uint8)
```

```python
#  Division
division = np.clip(img.astype(float) / 1.5, 0, 255).astype(np.uint8)
```

```python
# Step 8: Display All Results
plt.figure(figsize=(16, 5))

plt.subplot(1, 5, 1)
plt.imshow(img, cmap='gray')
plt.title("Normal Image")
plt.axis('off')

plt.subplot(1, 5, 2)
plt.imshow(addition, cmap='gray')
plt.title("Addition")
plt.axis('off')

plt.subplot(1, 5, 3)
plt.imshow(subtraction, cmap='gray')
plt.title("Subtraction")
plt.axis('off')

plt.subplot(1, 5, 4)
plt.imshow(multiplication, cmap='gray')
plt.title("Multiplication")
plt.axis('off')

plt.subplot(1, 5, 5)
plt.imshow(division, cmap='gray')
plt.title("Division")
plt.axis('off')

plt.tight_layout()
plt.show()
```
